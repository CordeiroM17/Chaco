```python
import sys, os
from pathlib import Path
# raíz del proyecto en el path
ROOT = Path.cwd()
if not (ROOT / 'src').exists() and (ROOT.parent / 'src').exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))
os.chdir(ROOT)

import pandas as pd
from IPython.display import Markdown, display
from src.scraping.transfermarkt_scraper import fetch_player, resumen_texto
import config
print('OK · delays SofaScore/Transfermarkt:', config.MIN_DELAY, '-', config.MAX_DELAY, 's')
```

## 1) Refuerzos a analizar

Editá este diccionario: **nombre → link de Transfermarkt**. Podés poner varios.
El de Matías Romero queda de ejemplo.

```python
REFUERZOS = {
    "Matías Romero": "https://www.transfermarkt.com.ar/matias-romero/profil/spieler/853464",
    "Leandro Fernandez": "https://www.transfermarkt.com.ar/leandro-fernandez/profil/spieler/355904",
    "Federico Moreno": "https://www.transfermarkt.com.ar/federico-moreno/profil/spieler/1139701",
    "Mauricio Tevez": "https://www.transfermarkt.com.ar/mauricio-tevez/profil/spieler/334097",
    # "Otro Refuerzo": "https://www.transfermarkt.com.ar/.../profil/spieler/XXXXX",
}
```

## 2) Bajar y parsear

Usa cache: si ya bajaste al jugador, no vuelve a la web. Para forzar una nueva
bajada, borrá su carpeta en `data/refuerzos/{id}/` o pasá `use_cache=False`.

```python
fichas = {}
for nombre, url in REFUERZOS.items():
    try:
        fichas[nombre] = fetch_player(url, use_cache=True)
        print(f'✓ {nombre}')
    except Exception as e:
        print(f'✗ {nombre}: {e}')
print(f'\n{len(fichas)} ficha(s) lista(s).')
```

## 3) Fichas

```python
for nombre, f in fichas.items():
    display(Markdown(resumen_texto(f)))
    display(Markdown('**Trayectoria (con vueltas):** ' + ' → '.join(f['trayectoria'])))
    display(pd.DataFrame(f['carrera']))
    display(Markdown('---'))
```

## 4) Guardar las fichas

Guarda un `.md` por jugador y un `.json` con todo, en `outputs/refuerzos/`.
Esto es lo que después alimenta la placa de Instagram del refuerzo.

```python
import json, re, unicodedata
OUT = Path('outputs/refuerzos'); OUT.mkdir(parents=True, exist_ok=True)
def _slug(s):
    s = unicodedata.normalize('NFKD', s).encode('ascii', 'ignore').decode()
    return re.sub(r'[^a-z0-9]+', '_', s.lower()).strip('_')
for nombre, f in fichas.items():
    slug = _slug(nombre)
    (OUT / f'{slug}.md').write_text(resumen_texto(f), encoding='utf-8')
    (OUT / f'{slug}.json').write_text(json.dumps(f, ensure_ascii=False, indent=2), encoding='utf-8')
    print('guardado:', OUT / f'{slug}.md')
```
